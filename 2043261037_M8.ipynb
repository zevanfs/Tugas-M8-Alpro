```python
#Nomor 6
pelamar = [
    ("Budi", 3.2, 5), 
    ("Siti", 3.6, 3), 
    ("Dodi", 3.8, 1)
]

print("====== HASIL EVALUASI PELAMAR ======")
for nama, ipk, semester in pelamar:
    if ipk >= 3.5 and semester > 2:
        status = "Diterima"
    else:
        status = "Ditolak"

    print(f"Mahasiswa : {nama:<5} | Status : {status}")
```

```python
#Nomor 7
karyawan = [
    ("Joko", "IT", 7000), 
    ("Rina", "HR", 5000), 
    ("Toni", "Data", 8000), 
    ("Lia", "IT", 6500)
]

total_gaji = 0

print("==== TOTAL GAJI KARYAWAN IT & DATA ====")
for nama, departemen, gaji in karyawan:
    if departemen == "IT" or departemen == "Data":
        total_gaji += gaji
        print(f"{nama:<20} : Rp.{gaji}")

print("-" * 40, "+")
print(f"Total gaji           : Rp.{total_gaji}")
```
